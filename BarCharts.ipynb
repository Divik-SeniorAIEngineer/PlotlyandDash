## Bar Charts

- A bar charts presents categorical data with rectangular bars with heights(or lengths) propotional to the values that they represent
- In general variables and data either represent measurements on some continious scale or they represent information about some categorical or discrete characteristics
- Using Bar Charts we can visualize categorical data.
- Typically the x-axis is the categoris and the y-axis is the count (number of occurrences) in each category
- However the y-axis can be any aggregation (count, sum, average,etc..)

```python
import numpy as np
import pandas as pd
import plotly.offline as pyo
import plotly.graph_objs as go
```

```python
df = pd.read_csv('Data/2018WinterOlympics.csv')
df
```

```python
data = [go.Bar(x=df['NOC'],y=df['Total'])]
layout = go.Layout(title='Medals')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
trace1 = go.Bar(x=df['NOC'],y=df['Gold'],name='Gold',marker={'color':'#FFD700'})
trace2 = go.Bar(x=df['NOC'],y=df['Silver'],name='Silver',marker={'color':'#9EA0A1'})
trace3 = go.Bar(x=df['NOC'],y=df['Bronze'],name='Bronze',marker={'color':'#CD7F32'})
data = [trace1,trace2,trace3]
layout = go.Layout(title='Medals')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
trace1 = go.Bar(x=df['NOC'],y=df['Gold'],name='Gold',marker={'color':'#FFD700'})
trace2 = go.Bar(x=df['NOC'],y=df['Silver'],name='Silver',marker={'color':'#9EA0A1'})
trace3 = go.Bar(x=df['NOC'],y=df['Bronze'],name='Bronze',marker={'color':'#CD7F32'})
data = [trace1,trace2,trace3]
layout = go.Layout(title='Medals',barmode='stack')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
df = pd.read_csv('Data/mocksurvey.csv')
df.head()
```

```python
df = df.set_index(df.columns[0])
df.index.name = 'Questions'
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
df.index
```

```python
data = [go.Bar(x=df.index,y=df[response],name=response) for response in df.columns]
```

```python
layout = go.Layout(title='Survey')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
layout = go.Layout(title='Survey',barmode='stack')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
data = [go.Bar(y=df.index,x=df[response],orientation='h',name=response) for response in df.columns]
```

```python
layout = go.Layout(title='Survey',barmode='stack')
fig = go.Figure(data,layout)
pyo.plot(fig)
```
