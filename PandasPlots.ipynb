## Bar Plot

Bars of different lengths show the distribution

- Categories on the x-axis
- Counts or the proportions on the y-axis

If there are too many categories, it could become confusing

- ##### Don't show averages per category using bar plot. Use boxplot or voilin plot in that case

```python
df.head()
```

```python
df.plot(x='bedrooms',y='price',kind='bar')
```

```python
df.groupby('bedrooms').count()
```

```python
df['bedrooms'].value_counts()
```

```python
df.groupby('bedrooms')['bedrooms'].count()
```

```python
bedrooms_count = df.groupby('bedrooms')['bedrooms'].count()
```

```python
bedrooms_count.plot()
```

```python
bedrooms_count.plot(kind='bar',title="Bedroom Counts",xlabel="Number of Bedrooms",ylabel="Count")
```

## Box Plot

- Box and Whisker Plot shows the summary data as three quartiles (Q1,Q2,Q3) and whiskers
- Multiple Box and Whisker plots shows the distribution of a continious variable(price) by a discrete variable(number of bedrooms)

```python
df.plot(y='price',kind='box')
```

```python
df.plot(y=['price','lotsize'],kind='box')
```

```python
df.plot(column='price',by='bedrooms',kind='box')
```

## Histogram and KDE(Kernel Density Estimate) Plot
KDE is a distribution plot which gives smooth curve

- Histograms show the distribution of a continious variable
- value on X axis
- Frequency on the y-axis

#### Bar plot has the discrete variable on the X axis, Histogram has a continious variable


### KDE or the Density Plot is used to visualize the distribution of observations (like Histogram)

```python
df.plot(y='price',kind='hist') # All houses between 20,000 and 40,000 bins each
```

```python
df.plot(y='price',kind='hist',bins=30) 
```

```python
df.plot(y='price',kind='kde') 
```

```python
df.plot(y='price',kind='kde',xlim=(0,50000)) 
```

## Scatter Plot

- Scatter plot is used to show the relationship between two continious variables
- Predictor on the X-axis
- Response on the Y-axis

```python
df.plot(x='lotsize',y='price',kind='scatter') # x=response y=predictor Based on the factor(response) analysing predictor
```
