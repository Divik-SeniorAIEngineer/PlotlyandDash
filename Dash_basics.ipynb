## Dash Layout
- Dash is for creating Dashboards purely in python
- we can interact with the dashboards
- Dash allows us to easily insert plotly graphs into the Dashboard

```python
import dash
import dash_core_components as dc
import dash_html_components  as html
import plotly.graph_objs as go
import numpy as np
import pandas as pd
```

```python
app = dash.Dash()
```

```python
app.layout = html.Div(children=[html.H1('Hello Dash!'),html.P('Welcome to Dash')])
```

```python
app.run_server()
```

```python
app.layout = html.Div(children=[html.H1('Hello Dash!'),html.P('Welcome to Dash'),
                               dc.Graph(id='example',figure={'data':[{'x':[1,2,3],'y':[4,1,2],'type':'bar','name':'SF'},{'x':[1,2,3],'y':[2,4,5],'type':'bar','name':'NYC'}],
                                                             'layout':{'title':'BAR Plots'}})
                               ])
```

```python
app.run_server()
```

## Customizing Basic Dash Layouts

```python
colors = {'background':'#111111','text':'#7FDBFF'}
app.layout = html.Div(children=[html.H1('Hello Dash!',style={'textAlign':'center','color':colors['text']}),html.P('Welcome to Dash',style={'textAlign':'center','color':colors['text']}),
                               dc.Graph(id='example',figure={'data':[{'x':[1,2,3],'y':[4,1,2],'type':'bar','name':'SF'},{'x':[1,2,3],'y':[2,4,5],'type':'bar','name':'NYC'}],
                                                             'layout':{'plot_bgcolor':colors['background'],'paper_bgcolor':colors['background'],'font':{'color':colors['text']},'title':'BAR Plots'}})], style={'backgroundColor':colors['background']})
```

```python
app.run_server()
```

## Converting Simple Plotly Plot to Dashboard with Dash

```python
np.random.seed(42)
random_x = np.random.randint(1,101,100)
random_y = np.random.randint(1,101,100)
```

```python
app.layout = html.Div([dcc.Graph(id='scatterplot',
                                figure={'data':[go.Scatter(x=random_x,y=random_y,mode='markers')],
                                        'layout':go.Layout(title='My Scatterplot')}
                                
                                )])
```

```python
app.run_server()
```

```python
app.layout = html.Div([dcc.Graph(id='scatterplot',
                                figure={'data':[go.Scatter(x=random_x,y=random_y,mode='markers',marker={'size':12,'color':'rgb(51,204,153)','symbol':'pentagon','line':{'width':2}})],
                                        'layout':go.Layout(title='My Scatterplot',xaxis={'title':'X title'})}
                                )])
```

```python
app.run_server()
```

```python

graph1 = dcc.Graph(id='scatterplot1',
                                figure={'data':[go.Scatter(x=random_x,y=random_y,mode='markers',marker={'size':12,'color':'rgb(51,204,153)','symbol':'pentagon','line':{'width':2}})],
                                        'layout':go.Layout(title='First Scatterplot',xaxis={'title':'X title'})}
                                )
graph2 = dcc.Graph(id='scatterplot2',
                                figure={'data':[go.Scatter(x=random_x,y=random_y,mode='markers',marker={'size':12,'color':'rgb(51,204,153)','symbol':'pentagon','line':{'width':2}})],
                                        'layout':go.Layout(title='Second Scatterplot',xaxis={'title':'X title'})}
                                )
app.layout = html.Div([graph1,graph2])

```

```python
app.run_server()
```
