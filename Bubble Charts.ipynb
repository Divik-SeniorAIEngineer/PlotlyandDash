## Bubble Charts

- Bubble charts are very similar to scatter plots , except we now coney a third variable's information through the size of the markers
- We can also continue to add variable information by coloring points based on a category

```python
import numpy as np
import pandas as pd
import plotly.offline as pyo
import plotly.graph_objs as go
```

```python
df = pd.read_csv('Data/mpg.csv')
df.head()
```

```python
data = [go.Scatter(x=df['horsepower'],y=df['mpg'],text=df['name'],mode='markers',marker=dict(size=2*df['cylinders']))]
```

```python
layout = go.Layout(title='Bubble Chart')
```

```python
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
data = [go.Scatter(x=df['horsepower'],y=df['mpg'],text=df['name'],mode='markers',marker=dict(size=df['weight']//100))]
```

```python
layout = go.Layout(title='Bubble Chart')
```

```python
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
data = [go.Scatter(x=df['horsepower'],y=df['mpg'],text=df['name'],mode='markers',marker=dict(size=df['weight']//100,color=df['cylinders']))]
```

```python
layout = go.Layout(title='Bubble Chart')
fig = go.Figure(data,layout)
pyo.plot(fig)
```

```python
df = pd.read_csv('Data/mpg.csv')
df.head()
```

```python
data = [go.Scatter(x=df['displacement'],y=df['acceleration'],text=df['name'],mode='markers',marker=dict(size=df['weight']//400))]
```

```python
layout = go.Layout(title='My Bubble',hovermode='closest')
```

```python
fig = go.Figure(data,layout)
pyo.plot(fig)
```
