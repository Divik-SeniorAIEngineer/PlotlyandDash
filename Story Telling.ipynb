#### Comparitive Visualizations
- Best used when you wish to show a series os values to be compared
- Effective at showing one or more series of values (bigger, smaller or similar)
- The comparitive charts fall into two categories:
- - Comparision between values
  - Comparision to the whole
- Comparision Between Values
- - Series of charts consists of the bar chart known as the column chart if bars are vertical
  - The horizontal bar chart or simply bar chart can be useful particularly when the bottom axis isn't time or you have particularly when the bottom axis isn't time or you have particularly long axis labels.
  - Grouped Columns chart variants of these if you are showing multiple series
- Comparision to whole
- - Show an audience how something compares to the whole
  - Ex: Numerous survey results and you want to display age breakdown of all survey respondents
- Ex: Bar Chart , Tree Map


#### TIME SERIES
- Common Visualization is one that shows change over time
- Seconds, hours, months, years decades and beyond
- Line Graph
- - Be careful not to plot too many lines
- Area Chart
- - It is like area chart but filled in(Space coloured)
  - Shows the total change over time


#### Correlation
- Show the relationship between two continious variables then you can select the Scatter plot
- For three continious variables use Bubble chart


#### Geographical Charts
- Not recommended when communicating with data
