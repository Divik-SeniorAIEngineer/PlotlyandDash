## Hover Over Data
- When you hover over Plotly graphs, information is displayed
- This information can be served internally using JSON and then parsed for specific information
- Every graph has a hoverData component porperty that can be accessed

```python
import dash
import dash_core_components as dc
import dash_html_components as html
from dash.dependencies import Input, Output , State
import plotly.graph_objs as go
import numpy as np
import pandas as pd
import base64 
import json
from numpy import random
```

```python
app = dash.Dash()
```

```python
df = pd.read_csv('Data/wheels.csv')
```

```python
app.layout = html.Div([
             html.Div(dc.Graph(id='wheels-plot',figure={'data':[go.Scatter(x=df['color'],y=df['wheels'],dy=1,mode='markers',marker={'size':15})],
                                                        'layout':go.Layout(title='Test',hovermode='closest')
                                                       })),
             html.Div(html.Pre(id='hover-data',style={'paddingTop':35}),
                     style={'width':'30%'}
                     ),
])
```

```python
@app.callback(Output('hover-data','children'),[Input('wheels-plot','hoverData')])
def callback_image(hover_data):
    return json.dumps(hover_data,indent=2)
    
    
```

```python
app.run_server()
```

```python
df = pd.read_csv('Data/wheels.csv')
```

```python
def encode_image(image_file):
    encoded = base64.b64encode(open(image_file,'rb').read())
    return 'data:image/png;base64,{}'.format(encoded.decode())
```

```python
app.layout = html.Div([
             html.Div(dc.Graph(id='wheels-plot',figure={'data':[go.Scatter(x=df['color'],y=df['wheels'],dy=1,mode='markers',marker={'size':15})],
                                                        'layout':go.Layout(title='Test',hovermode='closest')
                                                       })),
             html.Div([html.Img(id='hover-image',src='children',height=300)],
                     style={'width':'30%','height':'30%','paddingTop':35}
                     ),
])
```

```python
@app.callback(Output('hover-image','src'),[Input('wheels-plot','hoverData')])
def callback_image(hoverData):
    if hoverData is None:
        return ''
    wheel = hoverData['points'][0]['y'] 
    color = hoverData['points'][0]['x'] 
    path = 'data/images/' 
    return encode_image(path+df[(df['wheels']==wheel) & (df['color']==color)]['image'].values[0])
```

```python
app.run_server()
```

```python
app.layout = html.Div([
    html.Div([dc.Graph(id='plot',figure={'data':[go.Scatter(x=df['x'],y=df['y'],mode='markers')],
                         'layout':go.Layout(title="Scatterplot",hovermode='closest',dragmode='select')})],
            style={'width':'60%','display':'inline-block'}
            ),
    html.Div([html.H1(id='density',style={'paddingTop':50})],style={'width':'30%','display':'inline-block','verticalAlign':'top'})
  ])
```

```python
@app.callback(Output('density','children'),[Input('plot','selectedData')])
def find_density(selectedData):
    #Calculate the density of selectedData
    # No points selected
    if not selectedData or not selectedData.get('points'):
        return 'Select points to calculate density'

    points = selectedData['points']

    # Number of selected points
    pts = len(points)

    # Get x and y values from selected points
    x_values = [point['x'] for point in points]
    y_values = [point['y'] for point in points]

    # Calculate range
    x_range = max(x_values) - min(x_values)
    y_range = max(y_values) - min(y_values)

    # Calculate area
    area = x_range * y_range

    # Avoid division by zero
    if area == 0:
        return 'Density cannot be calculated for zero area'

    # Calculate density
    density = pts / area

    return 'Density = {:.2f}'.format(density)
    
```

```python
app.run_server()
```

## Updating Graph on Interactions

```python
app =  dash.Dash()
```

```python
df = pd.read_csv('Data/mpg.csv')
df.head()
df['year'] = random.randint(-4,5,len(df))*0.1 + df['model_year']
```

```python
df['year']
```

```python
app.layout = html.Div([
        html.Div([dc.Graph(id='mpg-scatter',figure={'data':[go.Scatter(
                            x=df['year']+1900,
                            y=df['mpg'],
                            text=df['name'],
                            hoverinfo='text+y+x',
                            mode='markers',
        )],'layout':go.Layout(title='MPG Data',xaxis={'title':'Model year'},yaxis={'title':'MPG'},
                             hovermode='closest')
                            }
                )],style={'width':'50%','display':'inline-block'}),
    html.Div([
        dc.Graph(id='mpg_line',figure={'data':[go.Scatter(x=[0,1],y=[0,1],mode='lines')],
                                       'layout':go.Layout(title='Acceleration',margin={'l':0})})
    ],style={'width':'20%','height':'50%','display':'inline-block'}),
    html.Div([
        dc.Markdown(id='mpg_stats')],style={'width':'20%','height':'50%','display':'inline-block'})
])
```

```python
@app.callback(Output('mpg_line','figure'),[Input('mpg-scatter','hoverData')])
def callback_graph(hoverData):
    if hoverData is None:
        return {}
    v_index= hoverData['points'][0]['pointIndex']
    figure = {'data':[go.Scatter(x=[0,1],y=[0,60/df.iloc[v_index]['acceleration']],mode='lines',line={'width':3*df.iloc[v_index]['cylinders']})],
              'layout':go.Layout(title=df.iloc[v_index]['name'],
               xaxis={'visible':False},
               yaxis={'visible':False,'range':[0,60/df['acceleration'].min()]},
               margin={'l':0},
               height=300  
        )}
    return figure
    
```

```python
@app.callback(Output('mpg_stats','children'),[Input('mpg-scatter','hoverData')])
def callback_stats(hoverData):
    if hoverData is None:
        return {}
    v_index = hoverData['points'][0]['pointNumber']
    stats = """
        {} cylinders
        {} cc displacement
        0 to 60mph in {} seconds
    """.format(df.iloc[v_index]['cylinders'],df.iloc[v_index]['displacement'],df.iloc[v_index]['acceleration'])
    return stats
```

```python
app.run_server()
```
