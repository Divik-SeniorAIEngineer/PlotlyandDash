## Displots - Distribution Plots
- Ditribution Plots or Displots, typically layer three plots on top of one another
- The first is a histogram where each data point is placed inside a bin of similar vallues.
- The second is a rug plot - marks are placed along the x-axis for every data point, which lets you see the distribution of values inside each bin
- Lastly, Distribution plots often include a "kernel density estimate" or KDE line that tries to describes the shape of the distribution

```python
import plotly.offline as pyo
import plotly.figure_factory as ff
import plotly.graph_objs as go
import pandas as pd
import numpy as np
```

```python
x = np.random.randn(1000)
hist_data = [x] 
```

```python
group_labels = ['distplot']
```

```python
fig = ff.create_distplot(hist_data,group_labels)
pyo.plot(fig)
```

```python
x1 = np.random.randn(200)-2
x2 = np.random.randn(200)
x3 = np.random.randn(200)+2
x4 = np.random.randn(200)+4
```

```python
hist_data = [x1,x2,x3,x4]
group_labels = ['X1','X2','X3','X4']
fig = ff.create_distplot(hist_data,group_labels)
pyo.plot(fig)
```

```python
hist_data = [x1,x2,x3,x4]
group_labels = ['X1','X2','X3','X4']
fig = ff.create_distplot(hist_data,group_labels,bin_size=[.2,.1,.3,.4])
pyo.plot(fig)
```
