## Bar Plot

```python
df = data('Housing')
df.head()
```

### Bar plot of number of houses versus the number of bedrooms

```python
bedrooms_count = df.groupby('bedrooms')['bedrooms'].count()
```

```python
bedrooms_count
```

```python
plt.bar(x=bedrooms_count.index,height=bedrooms_count,label='Number of Bedrooms')
plt.title("Houses by number of bedrooms")
plt.xlabel('Number of bedrooms')
plt.ylabel('Count')
plt.legend()
plt.grid(axis='y',alpha=0.2,ls='--')
```

#### Pareto Plot

```python
bedrooms_count = df.groupby('bedrooms')['bedrooms'].count()
bc_descending_order = bedrooms_count.sort_values(ascending=False)
bc_descending_order
```

```python
plt.bar(x=bc_descending_order.index.astype(str),height=bc_descending_order,label='Number of Bedrooms')
plt.title("Houses by number of bedrooms")
plt.xlabel('Number of bedrooms')
plt.ylabel('Count')
plt.legend()
plt.grid(axis='y',alpha=0.2,ls='--')
```

#### Histogram

```python
plt.bar(x=bc_descending_order.index.astype(str),height=bc_descending_order,label='Number of Bedrooms')
plt.plot(bc_descending_order.index.astype(str),np.cumsum(bc_descending_order),'o-r',alpha=0.5)
plt.axhline(0.80 * bc_descending_order.sum()) # Horizontal line at the 80 percent of the total no of houses
plt.title("Houses by number of bedrooms")
plt.xlabel('Number of bedrooms')
plt.ylabel('Count')
plt.legend(loc='upper left')
plt.grid(axis='y',alpha=0.2,ls='--')
```

```python
plt.barh(y=bc_descending_order.index.astype(str),width=bc_descending_order,label='Number of Bedrooms')
plt.title("Houses by number of bedrooms")
plt.ylabel('Number of bedrooms')
plt.xlabel('Count')
plt.legend(loc='upper left')
plt.grid(axis='y',alpha=0.2,ls='--')
```

## Subplots

Using subplot interface

```python
df = data('austres')
df.head()
```

```python
plt.plot(df['time'],df['austres'])
```

```python
df['austres'].pct_change()
```

```python
df['percent change'] = df['austres'].pct_change()
```

```python
df.head()
```

```python
plt.plot(df['time'],df['austres'])
plt.plot(df['time'],df['percent change'])
```

```python
fig, ax = plt.subplots()
fig, ax
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(10,5)
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(10,5)
ax.plot(df['time'],df['austres'])
ax.set_title('Australian Population')
ax.set_xlabel('Years')
ax.set_ylabel('Population (in thousands)')
fig.savefig('Aus.png',dpi=200)
```

```python
fig, (ax1,ax2)= plt.subplots(1,2,figsize=(10,5))
ax1.plot(df['time'],df['austres'])
ax1.set_title('Australian Population')
ax1.set_xlabel('Years')
ax1.set_ylabel('Population (in thousands)')
ax2.plot(df['time'],df['percent change'])
ax2.set_title('Australian+1000 Population')
ax2.set_xlabel('Years')
ax2.set_ylabel('Population (in thousands)')
fig.tight_layout()
fig.savefig('Aus.png',dpi=200)
```

```python
mean_pct_change = df['percent change'].mean()
```

```python
min_pct_change_y = df['percent change'].min()
min_pct_change_y
min_pct_change_x = df[df['percent change'] == min_pct_change_y ]['time'].iloc[0]
```

```python
fig, (ax1,ax2)= plt.subplots(1,2,figsize=(10,5))
ax1.plot(df['time'],df['austres'])
ax1.set_title('Australian Population')
ax1.set_xlabel('Years')
ax1.set_ylabel('Population (in thousands)')
ax2.plot(df['time'],df['percent change'])
ax2.set_title('Australian+1000 Population')
ax2.set_xlabel('Years')
ax2.set_ylabel('Population (in thousands)')

ax2.hlines(mean_pct_change,df['time'].min(),df['time'].max())
ax2.annotate('Mean % change', xy=(df['time'].min(),mean_pct_change-0.0002))

ax2.annotate('What happened here?',xy=(min_pct_change_x ,min_pct_change_y),xytext = (min_pct_change_x - 8,min_pct_change_y * 1.2),arrowprops={})

fig.tight_layout()
fig.savefig('Aus.png',dpi=200)
```
