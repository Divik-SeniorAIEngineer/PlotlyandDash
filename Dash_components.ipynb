## Dashboard Components
- Dash allows you to leverage previous knowledge of HTML and CSS to create very customized dashboards.
- Pick the relevant html component
- Adjust general CSS library
- An HTML Div element is a division. Section or block of the web app (dashboard)
- CSS allows for styling HTML elements
- Fonts, Colors, Borders etc...
- Dash uses dictionaries to pass in CSS style calls.

````python
import dash
import dash_core_components as dc
import dash_html_components  as html
import plotly.graph_objs as go
import numpy as np
import pandas as pd
````

````python
app = dash.Dash()
````

````python
app.layout = html.Div(["This is the outermost div!"],style={'color':'green','border':'2px green solid'})
````

````python
app.run_server()
````

````python
app.layout = html.Div(["This is the outermost div!",
                      html.Div(['This is an inner div!'],
                               style={'color':'red'})
                      ],style={'color':'green','border':'2px green solid'})
````

````python
app.run_server()
````

````python
app.layout = html.Div(["This is the outermost div!",
                       html.Div(['This is an inner div!'],
                               style={'color':'red'}),
                       html.Div('This is an inner inner div!',
                               style={'color':'pink'})
                      ],style={'color':'green','border':'2px green solid'})
````

````python
app.run_server()
````

````python
app.layout = html.Div(["This is the outermost div!",
                       html.Div(['This is an inner div!'],
                               style={'color':'red','border':'2px red solid'}),
                       html.Div('This is an inner inner div!',
                               style={'color':'pink'})
                      ],style={'color':'green','border':'2px green solid'})
````

````python
app.run_server()
````

## Core Components 
- Dash core components are more abstract (higher level) calls that allow you to quickly insert common components into your dashboard

````python
dc.__version__
````

````python
app.layout = html.Div([

    html.Label('Dropdown'),
    dc.Dropdown(options=[{'label':'New york city','value':'NYC'},{'label':'San Francisco','value':'SF'}]),
    html.Label('Slider'),
    dc.Slider(min=-10,max=10,step=0.5,value=0)
])
````

````python
app.run_server()
````

````python
app.layout = html.Div([

    html.Label('Dropdown'),
    dc.Dropdown(options=[{'label':'New york city','value':'NYC'},{'label':'San Francisco','value':'SF'}]),
    html.Label('Slider'),
    dc.Slider(min=-10,max=10,step=5,value=0,marks={i:i for i in range(-10,10,5)})
])
````

````python
app.run_server()
````

````python
app.layout = html.Div([

    html.Label('Dropdown'),
    dc.Dropdown(options=[{'label':'New york city','value':'NYC'},{'label':'San Francisco','value':'SF'}]),
    html.Label('Slider'),
    dc.Slider(min=-10,max=10,step=0.5,value=0,marks={i:i for i in range(-10,10,2)})
])
````

````python
app.run_server()
````

````python
app.layout = html.Div([

    html.Label('Dropdown'),
    dc.Dropdown(options=[{'label':'New york city','value':'NYC'},{'label':'San Francisco','value':'SF'}]),
    html.Label('Slider'),
    dc.Slider(min=-10,max=10,step=0.5,value=0,marks={i:i for i in range(-10,10,2)}),
    html.Label('Radio'),
    dc.RadioItems(options=[{'label':'New york city','value':'NYC'},{'label':'San Francisco','value':'SF'}],
                 value='SF'
                 )
])
````

````python
app.run_server()
````

## Markdown with Dash
- Dashboards can also display markdown text which 

````python
markdown_text = '''
# Data Analysis Report

## Introduction

This report contains a summary of the **data analysis** performed using Python.

### Dataset

The dataset contains the following columns:

- Name
- Age
- Gender
- Salary

## Python Example

```python
import pandas as pd

df = pd.read_csv("data.csv")
print(df.head())'''
````

````python
app.layout = html.Div([dc.Markdown(children=markdown_text)])
````

````python
app.run_server()
````
