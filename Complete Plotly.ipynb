### 1. ScatterPlot

### Scatter plots are drawn between to continous variables
### Problem :- We are going to draw a scatter plot between Batsman Avg(X axis) and
### Batsman Strike Rate(Y axis) of the top 50 batsman in IPL(All time)

##### Avg vs SR graph of Top 50 batsman(in terms of total runs)

##### Fetching a new dataframe with Top 50 batsman

```python
ipl.groupby('batter').sum()
```

```python
ipl.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(50).index
```

```python
top50
```

```python
new_ipl=ipl[ipl['batter'].isin(top50)]
```

```python
new_ipl.head()
```

```python
# Calculating SR
# SR=[(number of runs scored)/(number of balls played)]*100

runs=new_ipl.groupby('batter')['batsman_runs'].sum()
balls=new_ipl.groupby('batter')['batsman_runs'].count()

sr=(runs/balls)*100

#sr=sr.reset_index()
sr
```

```python
sr=sr.reset_index()
sr
```

```python
# Calculating Avg
# Avg=(Total number of Runs)/(Number of outs)

# Calculating number of outs for top 50 batsman
out=ipl[ipl['player_dismissed'].isin(top50)]

nouts=out['player_dismissed'].value_counts()

avg=runs/nouts

avg=avg.reset_index()
avg.rename(columns={'index':'batter',0:'avg'},inplace=True)

avg=avg.merge(sr,on='batter')
avg
```

```python
trace = go.Scatter(x=avg['avg'],y=avg['batsman_runs'],mode='markers',text=avg['batter'],marker={'color':'#99f66f','size':16})
```

```python
data =[trace]
layout = go.Layout({'title':'Batsman Avg Vs SR'},xaxis={'title':'Bastman Average'},yaxis={'title':'Batsman Strike Rate'})
fig = go.Figure(data=data,layout=layout)
pyo.plot(fig,filename='myfile.html')
```

### Bar Plot

#### Used to show the relationship between one categorical and one numerical


#### There are three types of Bar Graphs

- Nested Bar Graph
- Satcked Bar Graph
- Overlayed Bar Graph



```python
top10=ipl.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10).index.tolist()
```

```python
top10
```

```python
top10_df=ipl[ipl['batter'].isin(top10)]
top10_df.head()
```

```python
top10_score=top10_df.groupby('batter')['batsman_runs'].sum().reset_index()
top10_score
```

```python
trace = go.Bar(x=top10_score['batter'],y=top10_score['batsman_runs'])

data = [trace]

layout = go.Layout(title='Top 10 IPL batter',xaxis={'title':'batter'},yaxis={'title':'score'})

fig = go.Figure(data=data,layout=layout)

pyo.plot(fig)
```

```python
iw=top10_df.groupby(['batter','inning'])['batsman_runs'].sum().reset_index()
iw
```

```python
mask=iw['inning']==1
mask2=iw['inning']==2
```

```python
mask
```

```python
mask2
```

```python
one=iw[mask]
two=iw[mask2]
```

```python
one
```

```python
two
```

```python
one.rename(columns={'batsman_runs':'1st Innings'},inplace=True)
two.rename(columns={'batsman_runs':'2nd Innings'},inplace=True)

final=one.merge(two,on='batter')[['batter','1st Innings','2nd Innings']]

final
```

```python
# Nested (By Default)

trace1 = go.Bar(x=final['batter'],y=final['1st Innings'],name='1st Innings',marker={'color':'#00a65a'})
trace2 = go.Bar(x=final['batter'],y=final['2nd Innings'],name='2nd Innings',marker={'color':'#00a65a'})

data = [trace1,trace2]

layout = go.Layout(title='Inning Wise Scores',xaxis={'title':'batter'},yaxis={'title':'score'})

fig = go.Figure(data=data,layout=layout)

pyo.plot(fig)
```

```python
# Overlay

trace1 = go.Bar(x=final['batter'],y=final['1st Innings'],name='1st Innings',marker={'color':'#00a65a'})
trace2 = go.Bar(x=final['batter'],y=final['2nd Innings'],name='2nd Innings',marker={'color':'#99a65a'})

data = [trace1,trace2]

layout = go.Layout(title='Inning Wise Scores',xaxis={'title':'batter'},yaxis={'title':'score'},barmode='overlay')

fig = go.Figure(data=data,layout=layout)

pyo.plot(fig,filename='overlay.html')
```

```python
# Stack

trace1 = go.Bar(x=final['batter'],y=final['1st Innings'],name='1st Innings',marker={'color':'#00a65a'})
trace2 = go.Bar(x=final['batter'],y=final['2nd Innings'],name='2nd Innings',marker={'color':'#99a65a'})

data = [trace1,trace2]

layout = go.Layout(title='Inning Wise Scores',xaxis={'title':'batter'},yaxis={'title':'score'},barmode='stack')

fig = go.Figure(data=data,layout=layout)

pyo.plot(fig)
```

## Box Plot

## Histogram
- A histogram is a plot that lets you discover and show the underlying frequency distribution (shape) of a set of continious data

```python
x = delivery.groupby('batter')['batsman_runs'].count()>150
x = x[x].index.to_list()
x
```

```python
new = delivery[delivery['batter'].isin(x)]

runs = new.groupby('batter')['batsman_runs'].sum()
balls = new.groupby('batter')['batsman_runs'].count()
```

```python
runs
```

```python
balls
```

```python
sr = (runs/balls) * 100
sr
```

```python
sr = sr.reset_index()
sr
```

```python
trace = go.Histogram(x=sr['batsman_runs'],name="Strike run variation")
data = [trace]
layout = go.Layout(title='Strike Rate Analysis',xaxis={'title':'Strike Rates'})

fig = go.Figure(data=data,layout=layout)
pyo.plot(fig)
```

## Heatmaps
- Heatmap is a graphical representation of data 

```python
six=delivery[delivery['batsman_runs']==6]
six=six.groupby(['batting_team','over'])['batsman_runs'].count().reset_index()

six
```

```python
# Plot Heatmap

trace=go.Heatmap(x=six['batting_team'],y=six['over'],z=six['batsman_runs'])

data=[trace]

layout=go.Layout(title='Six Heatmap')

fig=go.Figure(data=data,layout=layout)

pyo.plot(fig)
```

```python
# Side by Side Heatmap

dots=delivery[delivery['batsman_runs']==0]
dots=dots.groupby(['batting_team','over'])['batsman_runs'].count().reset_index()

from plotly import tools

trace1=go.Heatmap(x=six['batting_team'],y=six['over'],
                 z=six['batsman_runs'].values.tolist())

trace2=go.Heatmap(x=dots['batting_team'],y=dots['over'],
                 z=dots['batsman_runs'].values.tolist())


fig=tools.make_subplots(rows=1,cols=2,subplot_titles=["6's","0's"], shared_yaxes=True)

fig.append_trace(trace1,1,1)
fig.append_trace(trace2,1,2)

pyo.plot(fig)
```
