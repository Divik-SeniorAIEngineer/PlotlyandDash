## Box Plots
- Box plots visualize the variation of a feature by deicting the continious numerical data through quartiles
- We can then seperate the data based on a categorcial feature to compare the continious feature based on category
- The box plot is a way of visually displaying the data distribution through their quartiles
- Quartiles separate the data into four equal parts
- Box plot is called a box and whisker plot
- The IQR is the interquartile range (the length of the filled in box Q3 - Q1)
- Outliers are values larger than Q3 by at least 1.5 times the IQR
- Outliers are values smaller than Q1 by at least 1.5 times the IQR

```python
import plotly.offline as pyo
import plotly.graph_objs as go
import pandas as pd
import numpy as np
```

```python
y = [1,14,14,15,16,18,18,19,19,20,20,23,24,26,27,27,28,29,33,54]
```

```python
data = [go.Box(y=y)]
```

```python
pyo.plot(data)
```

```python
data = [go.Box(y=y,boxpoints='all')]
pyo.plot(data)
```

```python
data = [go.Box(y=y,boxpoints='all',jitter=0.9)]
pyo.plot(data)
```

```python
data = [go.Box(y=y,boxpoints='all',jitter=0.9,pointpos=2.0)]
pyo.plot(data)
```

```python
data = [go.Box(y=y,boxpoints='all',jitter=0.1,pointpos=0)]
pyo.plot(data)
```

```python
data = [go.Box(y=y,boxpoints='outliers')]
pyo.plot(data)
```

```python
snodgrass = [.209,.205,.196,.210,.202,.207,.224,.223,.220,.201]
twain = [.225,.262,.217,.240,.230,.229,.235,.217]
```

```python
data = [go.Box(y=snodgrass,name='snodgrass'),go.Box(y=twain ,name='twain ')]
pyo.plot(data)
```
